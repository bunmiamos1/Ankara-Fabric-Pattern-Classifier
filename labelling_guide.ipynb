# ANKARA FABRIC PATTERN CLASSIFIER
## Pattern Classification and Labelling Guide

### 1. PROJECT OBJECTIVE

The objective of this project is to develop a machine learning model capable of identifying and classifying different pattern types found on Ankara fabrics.

The model should be able to:
- Identify all meaningful pattern categories present in a fabric.
- Recognise when a fabric contains more than one pattern category.
- Identify the dominant pattern or patterns in a fabric.
- Distinguish between dominant and secondary patterns.

**Important:** This project will use a multi-label classification approach. This means one fabric image can belong to multiple pattern classes rather than being restricted to just one category.

---

### 2. PROPOSED PATTERN CLASSES

The following six classes will be used:

1. Floral
2. Abstract
3. Geometric
4. Circular
5. Linear
6. Others

A particular fabric may contain elements belonging to two or more classes. Each meaningful pattern category should be recorded.

### 3. CLASS DEFINITIONS AND REQUIREMENTS

#### 3.1 FLORAL

A fabric should be classified as Floral if it contains recognisable or stylised botanical elements.

These may include:
- Any variation or representation of flowers.
- Flower petals, whether large or small.
- Leaves or different leaf structures.
- Trees, branches or root-like designs.
- Repeated floral arrangements.
- Stylised botanical illustrations.

The floral elements do not have to look exactly like real flowers. Artistic or abstract representations of flowers and leaves can still qualify if their botanical appearance is recognisable.

#### 3.2 ABSTRACT

A fabric should be classified as Abstract if it contains artistic, irregular or non-representational designs.

These may include:
- Irregular shapes and forms.
- Brushstroke-like patterns.
- Random artistic arrangements.
- Unstructured decorative shapes.
- Artistic designs that do not clearly represent recognisable objects or conventional geometric forms.

#### 3.3 GEOMETRIC

A fabric should be classified as Geometric if it contains identifiable geometric shapes or structured geometric arrangements.

These may include:
- Triangles.
- Squares and rectangles.
- Diamonds and rhombuses.
- Polygons.
- Angular shapes.
- Repeated symmetrical arrangements.
- Structured combinations of geometric figures.

#### 3.4 CIRCULAR

A fabric should be classified as Circular if it contains prominent circular or rounded design elements.

These may include:
- Full circles.
- Concentric circles.
- Dots and dotted arrangements.
- Rings.
- Rounded motifs.
- Repeated circular patterns.

Small incidental rounded shapes should not automatically qualify. The circular elements should form a meaningful part of the fabric's design.

#### 3.5 LINEAR

A fabric should be classified as Linear if it contains prominent line-based patterns.

These may include:
- Straight lines.
- Parallel lines.
- Stripes.
- Diagonal lines.
- Wavy lines.
- Zigzag patterns.
- Repeated linear arrangements.

#### 3.6 OTHERS

This category should be used for meaningful pattern designs that do not reasonably fit into any of the five categories above.

**Important:** Others should not be used simply because a pattern is difficult to identify. If the pattern is unclear or the annotator is unsure, the image should be marked for review.

---

### 4. MULTI-LABEL CLASSIFICATION REQUIREMENTS

A fabric may contain one or more of the six pattern categories.

For example:
- A fabric containing flowers and geometric shapes should be labelled Floral and Geometric.
- A fabric containing circles, stripes and abstract shapes should be labelled Circular, Linear and Abstract.
- A fabric containing only flowers should be labelled Floral.

Every meaningful pattern category present should be recorded, even if it is not the dominant pattern.

A pattern element must be visually meaningful to qualify as a category. Tiny incidental shapes that do not contribute meaningfully to the overall design should not automatically be labelled as a separate class.

### 5. DOMINANT AND SECONDARY PATTERNS

In addition to identifying all patterns present, each image must be assessed to determine which patterns are dominant and which are secondary.

#### 5.1 Dominant Patterns

A dominant pattern is a pattern category that has significant visual prominence in the overall fabric design.

Dominance should be assessed using:
- The size of the pattern elements.
- How frequently the pattern appears.
- Its visual contrast and prominence.
- Its contribution to the overall design composition.

A fabric may have:
- One dominant pattern.
- Two co-dominant patterns.
- Three or more co-dominant patterns, where genuinely justified.

**Multiple co-dominant patterns are allowed.**

For example, if Floral and Geometric patterns are equally prominent in a fabric, both should be recorded as dominant.

Do not force a single dominant category when two or more patterns are genuinely equally prominent.

#### 5.2 Secondary Patterns

Secondary patterns are visible pattern categories that contribute to the overall fabric design but are less prominent than the dominant pattern or patterns.

For example:

A fabric containing large floral arrangements and small circular dots may be labelled:
- Patterns present: Floral, Circular.
- Dominant patterns: Floral.
- Secondary patterns: Circular.

#### 5.3 Determining Dominance

Dominance should be based on the overall visual impression of the fabric, not simply the number of pattern elements or the percentage of space they occupy.

The annotator should consider the pattern's size, repetition, contrast and overall contribution to the design.

If it is difficult to determine which pattern is dominant, the image should be marked for review.

---

### 6. LABELLING EXAMPLES

| Fabric design | Patterns present | Dominant patterns | Secondary patterns |
|---|---|---|---|
| Large flowers with small dots | Floral, Circular | Floral | Circular |
| Equally prominent flowers and diamonds | Floral, Geometric | Floral, Geometric | None |
| Bold stripes with small flowers | Linear, Floral | Linear | Floral |
| Large circles and geometric shapes of equal prominence | Circular, Geometric | Circular, Geometric | None |
| Abstract brushstrokes with small circular motifs | Abstract, Circular | Abstract | Circular |
| Only floral patterns | Floral | Floral | None |

---

### 7. LABEL RECORDING FORMAT

Each image should have its own record.

The following information should be collected:

| Field | Description |
|---|---|
| image_name | Name of the image file |
| design_id | Identifier for the physical fabric design or source |
| patterns_present | All pattern categories identified |
| dominant_patterns | All dominant pattern categories |
| secondary_patterns | Pattern categories that are present but less prominent |
| colour | Main colour or colours of the fabric |
| background | Description of the image background |
| review_status | Whether the image has been reviewed and approved |

The six categories should always follow this order:

Floral, Abstract, Geometric, Circular, Linear, Others.

For machine learning, each image will eventually be represented using binary labels.

Example:

A fabric containing Floral, Geometric and Circular patterns, with Floral and Geometric being co-dominant:

- Patterns present: `[1, 0, 1, 1, 0, 0]`
- Dominant patterns: `[1, 0, 1, 0, 0, 0]`
- Secondary patterns: `[0, 0, 0, 1, 0, 0]`

Here:
- `1` means the pattern category is present.
- `0` means the pattern category is absent.

A dominant pattern must always be included in the patterns-present labels.

---

### 8. UNCERTAIN OR AMBIGUOUS IMAGES

Not every fabric will be easy to classify.

Where an image is unclear:
- Do not guess the pattern category.
- Do not automatically assign it to Others.
- Mark it as `needs_review`.
- Revisit the image before including it in the final labelled dataset.

If the image quality is too poor to identify the pattern, it should be excluded from training until a clearer image is available.

---

### 9. GENERAL LABELLING RULES

1. Label based on the visible fabric pattern, not the background or surrounding objects.
2. A single fabric can belong to multiple pattern classes.
3. Record every meaningful pattern category present.
4. Identify all dominant patterns, allowing co-dominance.
5. Record less prominent categories as secondary patterns.
6. Do not confuse colour variations with different pattern categories.
7. Do not use Others as a substitute for uncertainty.
8. Maintain consistency when labelling similar designs.
9. Mark unclear images for review instead of guessing.
10. Ensure dominant patterns are always included among the patterns present.

### 10. FINAL NOTE

The purpose of this guide is to establish a consistent labelling standard for the Ankara Fabric Pattern Classifier.

The dataset should accurately represent the complexity and diversity of Ankara fabric designs. Since some fabrics combine multiple design elements, the classification system must be flexible enough to recognise several pattern categories and identify more than one dominant pattern where necessary.

Consistency in labelling is essential because the quality of the dataset will directly influence the model's ability to recognise and classify Ankara fabric patterns accurately.
